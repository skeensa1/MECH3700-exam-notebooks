# **Heun's Method and Runge-Kutta (RK4)**
**Heun’s** method (Explicit Trapezoidal) improves upon Forward Euler by recognizing that a single slope at the beginning of a time interval ($t_n$) causes systematic overshoot or undershoot.
***
**Heun's Method in RK4 Terminology**
$$\begin{aligned} k_1 &= f(t_n, y_n) && \text{(Slope at start)} \\ k_2 &= f(t_n + \Delta t, y_n + \Delta t \cdot k_1) && \text{(Slope at predicted end)} \\ y_{n+1} &= y_n + \frac{\Delta t}{2}(k_1 + k_2) && \text{(Trapezoidal average)} \end{aligned}$$ $$$$

**The Limitation of Heun’s Method:**  

Heun only samples the interval boundaries ($t_n$ and $t_n + \Delta t$). If the underlying physics involves rapid acceleration, curvature, or non-linear state dependencies—such as our non-linear RC diode circuit or a non-linear pendulum—a linear slope average across the endpoints misses the behavior in the middle of the interval.  
***
**The Leap to RK4:**  
RK4 solves this by taking four strategic slope samples across the interval—one at the start, two at the midpoint, and one at the end—and combining them using Simpson’s Rule weights.  
***
The Geometric Intuition of RK4  
Instead of treating RK4 as a black-box formula, present the four stage slopes ($k_1, k_2, k_3, k_4$) as a trial-and-correct sequence across a single time step $\Delta t$: $$$$
1. $k_1$ (Initial Slope): The standard Forward Euler slope evaluated at the current state $(t_n, y_n)$.$$k_1 = f(t_n, y_n)$$
2. $k_2$ (Midpoint Trial Slope): Uses $k_1$ to step halfway across the interval ($\frac{\Delta t}{2}$) and evaluates the slope there.$$k_2 = f\left(t_n + \frac{\Delta t}{2}, y_n + \frac{\Delta t}{2} k_1\right)$$  
3. $k_3$ (Refined Midpoint Slope): Uses $k_2$ (a better midpoint estimate) to re-step halfway across the interval and re-evaluates the midpoint slope.$$k_3 = f\left(t_n + \frac{\Delta t}{2}, y_n + \frac{\Delta t}{2} k_2\right)$$  
4. $k_4$ (Endpoint Trial Slope): Uses $k_3$ to step all the way across the interval to $t_n + \Delta t$ and evaluates the slope at the end.$$k_4 = f(t_n + \Delta t, y_n + \Delta t \cdot k_3)$$ $$$$
***
The Simpson's Rule ConnectionThe magic of RK4 lies in how these four slopes are combined:$$y_{n+1} = y_n + \frac{\Delta t}{6} \left( k_1 + 2k_2 + 2k_3 + k_4 \right)$$  
Highlight the weighting factor $\frac{1}{6}(1 + 2 + 2 + 1) = 1$:  
- $k_1$ and $k_4$ (Endpoints): Given a weight of $1/6$ each.  
- $k_2$ and $k_3$ (Midpoints): Given a weight of $2/6 = 1/3$ each, because the midpoint derivative carries twice the weight in approximating the interval's trajectory.
